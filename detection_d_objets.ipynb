```python
from ultralytics import YOLO
import cv2

# Charger le modèle YOLO pré-entraîné
model = YOLO("yolov8n.pt")  # Modèle léger

# Charger une image
image = cv2.imread("image.png")

# Effectuer la détection
results = model(image)

# Afficher les résultats
results = results[0]
results.show()
```
