```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Charger l'image
image = cv2.imread('image.png', cv2.IMREAD_COLOR)  # Remplacer 'image.jpg' par le chemin de votre image

# Convertir l'image en niveaux de gris
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Appliquer un seuil pour la segmentation
# Ici, 127 est la valeur seuil. Les pixels supérieurs seront définis comme 255 (blanc),
# et ceux inférieurs à 127 seront définis comme 0 (noir)
_, thresholded_image = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# Affichage de l'image d'origine et de l'image segmentée
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))  # Convertir de BGR à RGB pour un affichage correct
plt.title('Image Originale')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(thresholded_image, cmap='gray')
plt.title('Image Segmentée')
plt.axis('off')

plt.show()
```
